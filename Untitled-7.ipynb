## Yêu cầu 1: Tiền xử lý dữ liệu

```python
def preprocess_corpus(text, remove_stopwords=False, min_word_length=2, min_word_freq=2, max_word_length=20):
    """
    Tiền xử lý văn bản - GIỮ LẠI stop words để câu có nghĩa tự nhiên
    
    Args:
        text: Văn bản gốc
        remove_stopwords: KHÔNG nên loại bỏ stop words (câu sẽ mất nghĩa)
        min_word_length: Độ dài tối thiểu của từ
        min_word_freq: Tần suất xuất hiện tối thiểu (loại bỏ từ hiếm/typo)
        max_word_length: Độ dài tối đa của từ (loại bỏ từ dính nhau)
    """
    # Wikipedia/HTML noise tokens - CHỈ loại bỏ những thứ này
    noise_tokens = {
        'http', 'https', 'www', 'html', 'htm', 'php', 'jpg', 'png', 'gif',
        'div', 'span', 'href', 'src', 'img', 'alt', 'class', 'style',
        'wiki', 'wikipedia', 'cite', 'ref', 'external', 'link', 'url',
        'nbsp', 'amp', 'quot', 'lt', 'gt', 'br', 'meta', 'script',
    }
    
    print("Preprocessing corpus...")
    text = text.lower()
    sentences = re.split(r'[.!?\n]+', text)
    sentences = [s.strip() for s in sentences if s.strip()]
    
    tokenized_sentences = []
    word_freq = Counter()
    filtered_count = 0
    
    for sentence in tqdm(sentences, desc="Processing"):
        sentence = re.sub(r'[^\w\s]', '', sentence)
        words = sentence.split()
        
        filtered_words = []
        for word in words:
            # Chỉ loại bỏ: noise tokens, số, từ quá ngắn/dài, mixed alphanumeric
            if (len(word) < min_word_length or 
                len(word) > max_word_length or
                word.isdigit() or
                any(char.isdigit() for char in word) or
                word in noise_tokens or
                any(token in word for token in ['http', 'www'])):
                filtered_count += 1
                continue
                
            filtered_words.append(word)
            word_freq[word] += 1
        
        if len(filtered_words) >= 2:
            tokenized_sentences.append(filtered_words)
    
    # Lọc từ hiếm
    valid_words = {word for word, freq in word_freq.items() if freq >= min_word_freq}
    filtered_sentences = [
        [w for w in sent if w in valid_words] 
        for sent in tokenized_sentences
    ]
    filtered_sentences = [s for s in filtered_sentences if len(s) >= 2]
    
    # Loại bỏ duplicate
    dedup_sentences = []
    for sentence in filtered_sentences:
        deduped = [word for i, word in enumerate(sentence) 
                   if i == 0 or word != sentence[i-1]]
        if len(deduped) >= 2:
            dedup_sentences.append(deduped)
    
    # Loại bỏ câu trùng lặp
    seen = set()
    unique_sentences = []
    for sentence in dedup_sentences:
        key = ' '.join(sentence)
        if key not in seen:
            seen.add(key)
            unique_sentences.append(sentence)
    
    # Summary
    total_words = sum(len(s) for s in unique_sentences)
    unique_words = len(set(w for s in unique_sentences for w in s))
    
    print(f"\n{'='*70}")
    print("PREPROCESSING SUMMARY")
    print(f"{'='*70}")
    print(f"Sentences: {len(sentences):,} → {len(unique_sentences):,}")
    print(f"Unique words: {unique_words:,} | Avg length: {total_words/len(unique_sentences):.1f} words")
    print(f"Filtered tokens: {filtered_count:,} (noise/numbers/length)")
    print(f"{'='*70}\n")
    
    return unique_sentences

# Tiền xử lý - GIỮ stop words để câu có nghĩa
tokenized_corpus = preprocess_corpus(
    corpus_text,
    remove_stopwords=False,  # GIỮ stop words
    min_word_length=2,
    min_word_freq=2,
    max_word_length=20
)
```

## Yêu cầu 2: Xây dựng Word Graph (Bigram Model)

```python
def build_word_graph(tokenized_sentences):
    """Xây dựng bigram graph với progress bar, loại bỏ bigram trùng lặp"""
    word_graph = defaultdict(set)  # Dùng set thay vì list để tự động loại bỏ trùng lặp
    
    print("Building word graph...")
    total_bigrams = 0
    for sentence in tqdm(tokenized_sentences, desc="Building graph", unit="sent"):
        for i in range(len(sentence) - 1):
            word_graph[sentence[i]].add(sentence[i + 1])  # add vào set
            total_bigrams += 1
    
    # Convert set về list để tương thích với code sau
    word_graph = {word: list(neighbors) for word, neighbors in word_graph.items()}
    
    nodes = len(word_graph)
    unique_edges = sum(len(neighbors) for neighbors in word_graph.values())
    
    print(f"✓ Word Graph: {nodes:,} nodes, {unique_edges:,} unique edges")
    print(f"  - Total bigrams processed: {total_bigrams:,}")
    print(f"  - Duplicate bigrams removed: {total_bigrams - unique_edges:,}")
    print(f"  - Deduplication rate: {(1 - unique_edges/total_bigrams)*100:.1f}%")
    
    return word_graph

word_graph = build_word_graph(tokenized_corpus)
```

```python
def analyze_graph_structure(word_graph):
    """Phân tích cấu trúc chi tiết của word graph"""
    
    # Tính toán metrics
    all_neighbors = [neighbor for neighbors in word_graph.values() for neighbor in neighbors]
    neighbor_counts = Counter(all_neighbors)
    
    out_degrees = {word: len(neighbors) for word, neighbors in word_graph.items()}
    in_degrees = dict(neighbor_counts)
    
    print(f"\n{'='*70}")
    print("DETAILED GRAPH ANALYSIS")
    print(f"{'='*70}")
    
    # Overall stats
    print(f"\nOverall Statistics:")
    print(f"  Total nodes: {len(word_graph):,}")
    print(f"  Total edges: {sum(len(v) for v in word_graph.values()):,}")
    print(f"  Avg out-degree: {sum(out_degrees.values()) / len(out_degrees):.2f}")
    print(f"  Avg in-degree: {sum(in_degrees.values()) / len(in_degrees):.2f}")
    
    # Top words by out-degree (words that lead to many other words)
    print(f"\nTop 5 words with most outgoing connections:")
    for word, degree in sorted(out_degrees.items(), key=lambda x: x[1], reverse=True)[:5]:
        next_words = word_graph[word][:5]
        print(f"  '{word}' → {degree} words (e.g., {', '.join(next_words)})")
    
    # Top words by in-degree (words that are followed by many other words)
    print(f"\nTop 5 most popular target words:")
    for word, degree in sorted(in_degrees.items(), key=lambda x: x[1], reverse=True)[:5]:
        print(f"  '{word}' ← {degree} incoming connections")
    
    # Isolated and terminal nodes
    isolated = [w for w in word_graph if len(word_graph[w]) == 0]
    print(f"\nTerminal nodes (no outgoing edges): {len(isolated)}")
    if len(isolated) <= 10:
        print(f"  Examples: {', '.join(isolated[:10])}")
    
    # Degree distribution
    degree_dist = Counter(out_degrees.values())
    print(f"\nOut-Degree Distribution:")
    for degree in sorted(degree_dist.keys())[:10]:
        print(f"  Degree {degree}: {degree_dist[degree]} words")
    
    return {
        'out_degrees': out_degrees,
        'in_degrees': in_degrees,
        'degree_distribution': degree_dist
    }

graph_stats = analyze_graph_structure(word_graph)
```

```python
import networkx as nx
import matplotlib.pyplot as plt
from collections import Counter

def visualize_word_graph_subset(word_graph, max_nodes=50, min_connections=3):
    """Visualize một phần word graph với các từ có nhiều kết nối nhất"""
    
    # Lọc các từ có nhiều kết nối
    node_degrees = {word: len(neighbors) for word, neighbors in word_graph.items()}
    top_nodes = [word for word, degree in sorted(node_degrees.items(), 
                                                  key=lambda x: x[1], 
                                                  reverse=True)[:max_nodes] 
                 if degree >= min_connections]
    
    # Tạo subgraph với progress bar
    G = nx.DiGraph()
    print(f"Building visualization graph...")
    for word in tqdm(top_nodes, desc="Adding nodes", unit="node"):
        for next_word in word_graph[word]:
            if next_word in top_nodes:
                if G.has_edge(word, next_word):
                    G[word][next_word]['weight'] += 1
                else:
                    G.add_edge(word, next_word, weight=1)
    
    print(f"✓ Subgraph: {G.number_of_nodes()} nodes, {G.number_of_edges()} edges")
    
    # Visualization với Circular Layout + Manual Adjustment
    plt.figure(figsize=(24, 20))
    print("Computing optimized layout...")
    
    # Node degrees
    node_degrees_in_G = dict(G.degree())
    
    # Sử dụng Circular Layout làm base - đảm bảo nodes dàn đều
    pos_circular = nx.circular_layout(G, scale=2.0)
    
    # Kết hợp với Spring Layout nhưng giữ một số nodes cố định
    pos = nx.spring_layout(
        G,
        pos=pos_circular,  # Bắt đầu từ circular
        k=8,  # Khoảng cách lớn giữa nodes
        iterations=200,  # Nhiều iterations để ổn định
        scale=3.5,  # Scale lớn hơn
        seed=42
    )
    
    # Node size và color theo degree
    node_sizes = [600 + node_degrees_in_G[node] * 25 for node in G.nodes()]
    node_colors = [node_degrees_in_G[node] for node in G.nodes()]
    
    # Edge width theo weight
    edges = G.edges()
    weights = [G[u][v]['weight'] for u, v in edges]
    
    # Vẽ graph với style đẹp
    print("Rendering graph...")
    
    # Vẽ edges trước (ở dưới)
    nx.draw_networkx_edges(
        G, pos, 
        width=[w*0.2 for w in weights], 
        alpha=0.15, 
        edge_color='#888888', 
        arrows=True, 
        arrowsize=15, 
        arrowstyle='->',
        connectionstyle='arc3,rad=0.2',  # Cong nhiều hơn để tránh chồng chéo
        node_size=node_sizes
    )
    
    # Vẽ nodes
    nodes = nx.draw_networkx_nodes(
        G, pos, 
        node_size=node_sizes, 
        node_color=node_colors,
        cmap=plt.cm.viridis,
        alpha=0.9,
        vmin=min(node_colors),
        vmax=max(node_colors),
        edgecolors='white',
        linewidths=2.5
    )
    
    # Vẽ labels với background
    for node, (x, y) in pos.items():
        plt.text(
            x, y, node,
            fontsize=11,
            fontweight='bold',
            color='white',
            ha='center',
            va='center',
            bbox=dict(
                boxstyle='round,pad=0.4',
                facecolor='#2d2d2d',
                edgecolor='white',
                alpha=0.85,
                linewidth=1.5
            ),
            zorder=10
        )
    
    # Colorbar
    cbar = plt.colorbar(nodes, label='Node Degree', shrink=0.7, pad=0.02)
    cbar.ax.tick_params(labelsize=11)
    
    # Title
    plt.title(
        f"Word Graph Visualization - Bigram Network\n" +
        f"{len(top_nodes)} Most Connected Words | Circular + Spring Layout",
        fontsize=20, fontweight='bold', pad=30, color='#2d2d2d'
    )
    
    plt.axis('off')
    plt.tight_layout()
    plt.show()
    
    # Statistics
    print(f"\n{'='*60}")
    print("GRAPH STATISTICS")
    print(f"{'='*60}")
    print(f"Nodes: {G.number_of_nodes()}")
    print(f"Edges: {G.number_of_edges()}")
    print(f"Avg degree: {sum(dict(G.degree()).values()) / G.number_of_nodes():.2f}")
    print(f"Density: {nx.density(G):.4f}")
    
    # Top connected words
    degrees = dict(G.degree())
    top_5 = sorted(degrees.items(), key=lambda x: x[1], reverse=True)[:5]
    print(f"\nTop 5 most connected words:")
    for word, degree in top_5:
        print(f"  '{word}': {degree} connections")
    
    return G

# Visualize word graph
G = visualize_word_graph_subset(word_graph, max_nodes=50, min_connections=3)
```

## Yêu cầu 3: Thuật toán sinh câu (chống lặp vô hạn)

```python
def generate_phrase_from_word(start_word, word_graph):
    """Sinh cụm từ với cơ chế chống lặp vô hạn"""
    if start_word not in word_graph:
        return start_word
    
    phrase = [start_word]
    visited = {start_word}
    current_word = start_word
    
    while True:
        unvisited_next = [w for w in word_graph.get(current_word, []) if w not in visited]
        if not unvisited_next:
            break
        next_word = unvisited_next[0]
        phrase.append(next_word)
        visited.add(next_word)
        current_word = next_word
    
    return ' '.join(phrase)

def generate_all_phrases(word_graph):
    """Sinh tất cả cụm từ từ mọi từ khởi đầu với progress bar"""
    all_words = list(word_graph.keys())
    print(f"Generating phrases from {len(all_words):,} words...")
    
    all_phrases = []
    for w in tqdm(all_words, desc="Generating phrases", unit="word"):
        all_phrases.append(generate_phrase_from_word(w, word_graph))
    
    phrase_lengths = [len(p.split()) for p in all_phrases]
    
    print(f"✓ Generated: {len(all_phrases):,} phrases, {len(set(all_phrases)):,} unique")
    print(f"  Length: min={min(phrase_lengths)}, max={max(phrase_lengths)}, avg={sum(phrase_lengths)/len(phrase_lengths):.1f}")
    
    return all_phrases

generated_phrases = generate_all_phrases(word_graph)
```

## Yêu cầu 4: Xếp hạng và xuất Top 10

```python
def rank_and_display_top_phrases(phrases, top_n=10):
    """Xếp hạng và hiển thị top N dài nhất"""
    unique_phrases = list(set(phrases))
    phrases_with_length = [(p, len(p.split())) for p in unique_phrases]
    phrases_with_length.sort(key=lambda x: x[1], reverse=True)
    
    print(f"\n{'='*70}")
    print(f"TOP {top_n} LONGEST PHRASES")
    print(f"{'='*70}")
    
    for i, (phrase, word_count) in enumerate(phrases_with_length[:top_n], 1):
        print(f"\n#{i} ({word_count} words): {phrase}")
    
    all_lengths = [length for _, length in phrases_with_length]
    print(f"\n{'='*70}")
    print(f"Stats: longest={max(all_lengths)}, shortest={min(all_lengths)}, avg={sum(all_lengths)/len(all_lengths):.1f}")
    print(f"{'='*70}")
    
    return phrases_with_length[:top_n]

top_10_phrases = rank_and_display_top_phrases(generated_phrases, top_n=10)
```
